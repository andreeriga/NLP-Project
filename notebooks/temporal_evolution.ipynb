```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_turns = pd.read_csv('../data/batch_runs/turns.csv')

df_turns['sentiment_compound'] = pd.to_numeric(df_turns['sentiment_compound'], errors='coerce')

fig, axes = plt.subplots(2, 1, figsize=(14, 12))

sns.lineplot(
    data=df_turns, 
    x='round_number', 
    y='sentiment_compound', 
    hue='speaker', 
    ax=axes[0], 
    marker='o', 
    linewidth=2,
    errorbar=None
)
axes[0].set_title('Average Sentiment Evolution per Round')
axes[0].set_xlabel('Round')
axes[0].set_ylabel('Sentiment Compound')
axes[0].grid(True, alpha=0.4)

sns.histplot(
    data=df_turns, 
    x='round_number', 
    hue='dominant_tactic', 
    multiple='stack', 
    ax=axes[1], 
    palette='Set2',
    discrete=True, 
    linewidth=0.3,
    alpha=0.8,
    edgecolor='white',
)
axes[1].set_title('Dominant Tactics Frequency per Round')
axes[1].set_xlabel('Round')
axes[1].set_ylabel('Tactics Count')

plt.tight_layout()
plt.savefig('../evaluation/turns_analysis.png', dpi=300)
plt.show()
```
